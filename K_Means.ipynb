```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

#  2D Data Points
X, _ = make_blobs(n_samples=400, centers=4, cluster_std=0.60, random_state=42)

# K-Means
k = 4
kmeans = KMeans(n_clusters=k, n_init=10, random_state=42)
y_kmeans = kmeans.fit_predict(X)
centroids = kmeans.cluster_centers_

#  Result
plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], s=40, color="gray", alpha=0.7)
plt.title("Original Synthetic Data Points")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=40, cmap="viridis", alpha=0.7)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    s=200,
    marker="X",
    label="Centroids",
)
plt.title(f"K-Means Clustering (K={k})")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
